```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
```

```python
#create a line plot 

from turtle import color


plt.plot(x,y,color='red')
plt.xlabel('x_Axis')
plt.ylabel('y-axis')
plt.title('Basic Line plot')
plt.show()
```

```python
#create customized line plot 


plt.grid(True)
plt.plot(x,y,color='red',linestyle='--',marker='o',linewidth=3,markersize=9)
```

```python
#Multiple plots 
#sample data 

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title("Plot 1")
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(y1,x,color='orange')
plt.title('figure 1')

plt.subplot(2,2,2)
plt.plot(x,y1,color='green')
plt.title('figure 2')

plt.subplot(2,2,3)
plt.plot(x,y2,color='red')
plt.title('figure 3')

plt.subplot(2,2,4)
plt.plot(y2,x,color='yellow')
plt.title('figure 4')
```

```python
categories=['A','B','C','D','E']
values = [ 10 , 20 ,30,40 ,50]

plt.bar(categories,values,color='purple')
```

#Histograms 

Histograms are mainly used to show the frequencey of data .
They divide data into bins and count the frequency of data points inside each bin 

```python
data = [1,1,2,3,1,2,3,3,3,3,2,4,4,4,1,1,5,6,5,7,6,8]

plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
#scatter plot 

x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.scatter(x,y,color='blue',marker='x')
```

```python
labels=['A','B','C','D']
sizes=[30,20,40,10]
colors=['gold','yellowgreen','lightcoral','lightskyblue']
explode=(0.2,0,0,0)


plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct="%1.1f%%")
```
