### Assignment 4: Data Aggregation

1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.
2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
import pandas as pd
import numpy as np

# Number of rows
n = 10

# Create DataFrame
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=n),
    'Value': np.random.randint(1, 100, size=n)
})

print(df)
```

```python
df.groupby('Category')['Value'].agg(['sum','mean'])
```

```python
import pandas as pd
import numpy as np

# First DataFrame
df1 = pd.DataFrame({
    'A': np.random.randint(1, 50, 3),
    'B': np.random.randint(1, 50, 3)
})

# Second DataFrame (different columns)
df2 = pd.DataFrame({
    'C': np.random.randint(50, 100, 3),
    'D': np.random.randint(50, 100, 3)
})

print("DataFrame 1:\n", df1)
print("\nDataFrame 2:\n", df2)

# 🔹 Concatenate along rows (axis=0)
concat_rows = pd.concat([df1, df2], axis=0)
print("\nConcatenated along rows:\n", concat_rows)

# 🔹 Concatenate along columns (axis=1)
concat_cols = pd.concat([df1, df2], axis=1)
print("\nConcatenated along columns:\n", concat_cols)
```

```python
import pandas as pd
import numpy as np

# Create daily date range for 3 months
dates = pd.date_range(start='2025-01-01', periods=90, freq='D')

# Create DataFrame with datetime index
df = pd.DataFrame(
    {'Value': np.random.randint(10, 100, size=90)},
    index=dates
)

print("Original Daily Data:\n", df.head())

# Resample to monthly frequency and compute mean
monthly_mean = df.resample('M').mean()

print("\nMonthly Mean:\n", monthly_mean)
```
