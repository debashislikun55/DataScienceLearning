 pandas is data manipulation library in python , widely used for data analysis and data cleaning 
 It provides two primary data structures : Series and data frame 

 Series : One dimensional array like objects

 
  
 Dataframe : Two dimensional , size mutable and heterogeneous data structure with labeled axes (rows and columns )

```python
### Merging and joining Dataframes
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,how="inner",on="Key")
```

```python
pd.merge(df1,df2,how="left",on="Key")
```
