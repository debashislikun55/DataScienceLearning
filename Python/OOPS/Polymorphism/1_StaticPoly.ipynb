```python
class implement:
    def add(a,b):
        return a+b
    def add(a,b,c):
        return a+b+c

imp=implement
sum=imp.add(2,3)
```

```python
class Implementation:
    def add(*args):
        return sum(args)
    
imp1=Implementation
res=imp1.add(1,2,3)
print(res)
```

```python
#Operator overloading 

class Vector:

    def __init__(self,x,y):
        self.x=x
        self.y=y
        
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    
    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    
    def __mul__(self,other):
        return Vector(self.x * other+self.y*other)
    
    def __eq__(self,other):
        return self.x==other.x and self.y==other.y
    


v1=Vector(2,3)
v2=Vector(3,4)

print(v1.__add__(v2))
print(v1.__sub__(v2))

        
```
