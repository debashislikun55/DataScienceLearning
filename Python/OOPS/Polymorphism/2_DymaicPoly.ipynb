```python
class Animal:
    def Bark(self):
        print("Animal Barking")

class Dog(Animal):
    def Bark(self):
        super().Bark()
        print("Dog is Barking")

Dog1=Dog()
Dog1.Bark()

        
```

```python
from abc import ABC,abstractmethod

@abstractmethod
class Shape(ABC):
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def area(self):
        return self.a*self.b
```

```python

rect = Rectangle(3, 4)
```

```python
print(rect.area())
```
