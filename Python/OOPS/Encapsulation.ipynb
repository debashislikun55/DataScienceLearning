```python
#Acess private data members using public getter and setter method 



class Employee:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
        
    #getter 

    def get_age(self):

        return self.__age,self.__name
    
    def set_age(self,value):
        if(value<0):
            print("Age can not be negative ")
        else :
            self.__age=value

    

    def greet(self):
        print(f"Hello {self.__name}")

emp1= Employee("Debashis",25) # While setting the value it applies name gambling automatically 

#But while doing outside access it does not do that 
```

```python
#Now access directly outside class using the public getter and setter method 

emp1.get_age()
```

```python
#Acess private data members using getter and setter properties 

class AgeError(Exception):
    pass


class Employee:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
        
    #getter  setter properties 
    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self,value):
        if(value<0):
            raise AgeError("Age Can not be negative")
        self.__age=value


    def greet(self):
        print(f"Hello {self.__name}")

emp1= Employee("Debashis",25) # While setting the value it applies name gambling automatically 

#But while doing outside access it does not do that 
```

```python
emp1.age
```

```python
emp1.age=28
```

```python
emp1.age
```
