```python
from abc import ABC,abstractmethod
```

```python
class Vehicle(ABC):

    @abstractmethod
    def cardetails(self):
        pass

    def drive(self):
        print("The vehicle is used for driving ")
```

```python
vh1=Vehicle()
```

```python
class Car(Vehicle):
    def cardetails(self):
        print("Car started ")
```

```python
car1=Car()
```

```python
car1.cardetails()
```

```python
class Dog:
    def sound(self):
        print("Bark")

class Human:
    def sound(self):
        print("Hello")

def make_sound(obj):
    obj.sound()
```

```python
make_sound(Dog())
make_sound(Human())
```
