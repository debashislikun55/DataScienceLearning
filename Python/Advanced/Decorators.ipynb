```python
def main_welcome(func):
    def subwelcome():
        print("Welcome to these course")
        func("Welcome to tutorial")
        print("Please learn these concepts")
    return subwelcome()
```

```python
main_welcome(print)
```

```python
def my_decorator(func):
    def wrapper():
        print("something is happening before the funtion is called")
        func()
        print("Something is happening after the function is called")
    return wrapper
```

```python
@my_decorator
def sayhello():
    print("Hello")
```

```python
sayhello()
```
